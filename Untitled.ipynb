```python
import pandas as pd
import ipaddress

# 加载数据集，指定编码为 'ISO-8859-1' 或 'latin1'
file_path = '1.csv'
df = pd.read_csv(file_path, encoding='latin1')

# 定义函数，将IP地址转换为整数
def ip_to_int(ip):
    try:
        return int(ipaddress.IPv4Address(ip))
    except ipaddress.AddressValueError:
        return None  # 处理无效的IP地址

# 将 srcip 和 dstip 列中的IP地址转换为整数
df['srcip_int'] = df['srcip'].apply(ip_to_int)
df['dstip_int'] = df['dstip'].apply(ip_to_int)

# 查看转换后的数据
print(df[['srcip', 'srcip_int', 'dstip', 'dstip_int']].head())

# 保存结果到一个新的CSV文件
df.to_csv('converted_ip_dataset.csv', index=False)
```
